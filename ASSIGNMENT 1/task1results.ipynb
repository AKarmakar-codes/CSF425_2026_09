```python
evaluation = evaluate_checkpoint(CHECKPOINT, MANIFEST, split='test', output_dir=OUTPUT)
summary = summarize_prediction_file(OUTPUT / 'predictions.json')
assert summary['argmax'] == evaluation['argmax']
for policy in ['argmax', 'frozen_threshold']:
    print(policy, {k: summary[policy][k] for k in ['count','accuracy','P_macro','R_macro','F1_macro','S_cls','coverage','rejected_count']})
print('Zero-support classes (0-based):', summary['argmax']['zero_support_classes'])
```
